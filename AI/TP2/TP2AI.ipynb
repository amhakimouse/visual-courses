```python
y_pred = a_mc * x + b_mc
plt.scatter(x, y, label="Données")
plt.plot(x, y_pred, color="red", label=f"y = {a_mc:.2f}x + {b_mc:.2f}")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
a, b = 0.0, 0.0
alpha = 0.0001
n_iter = 1000
cost_history = []
```

```python
for i in range(n_iter):
    y_pred = a * x + b
    error = y - y_pred
    da = -(1 / n) * np.sum(x * error)
    db = -(1 / n) * np.sum(error)
    a = a - alpha * da
    b = b - alpha * db
    J = (1 / (2 * n)) * np.sum(error ** 2)
    cost_history.append(J)
```

```python
plt.plot(cost_history)
plt.xlabel("Itération")
plt.ylabel("J(a, b)")
plt.title("Évolution de la fonction de coût")
plt.grid(True)
plt.show()
```
